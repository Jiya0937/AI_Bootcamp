```python
correlation = df.corr(numeric_only=True)

print(correlation)

fraud_by_location = df.groupby("claim_location")["is_fraudulent"].mean()

print(fraud_by_location)
```

```python
df['policyholder_id'] = df['policyholder_id'].fillna('Unknown').replace('', 'Unknown')
df["claim_amount"] = df["claim_amount"].fillna(df["claim_amount"].mean)
```

```python
#handle missing values

df["claim_date"] = pd.to_datetime(df["claim_date"], errors="coerce")
df["claim_date"] = df["claim_date"].ffill()
df["claim_location"] = df["claim_location"].fillna("Unknown")
df["submission_channel"] = df["submission_channel"].fillna("Unknown")
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

# Feature Engineering

```python
df["claim_year"] = df["claim_date"].dt.year
df["claim_month"] = df["claim_date"].dt.month
df["claim_day"] = df["claim_date"].dt.day
df["claim_dayofweek"] = df["claim_date"].dt.dayofweek
```
